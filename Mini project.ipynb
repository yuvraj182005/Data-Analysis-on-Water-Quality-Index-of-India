```python
start = 1
end = 1779
station = df.iloc[start:end, 0]
location = df.iloc[start:end ,1]
state = df.iloc[start:end, 2]
do = df.iloc[start:end, 4].astype(np.float64)

value=0

ph = df.iloc[ start:end, 5]  
co = df.iloc [start:end, 6].astype(np.float64)   
  
tc = df.iloc[1:end, 10].astype(np.float64)
fc = df.iloc[1:end, 10].astype(np.float64)

bod = df.iloc[start:end, 7].astype(np.float64)
na = df.iloc[start:end, 8].astype(np.float64)
na.dtype
```

```python
df
```

```python
df = pd.concat([station,location,state,do,ph,co,bod,na,fc,tc], axis=1)

df.columns = ['station','location','state','do','ph','co','bod','na','fc','tc']
```

```python
#calculation of total coliform
df['nco']=df.tc.apply(lambda x:(100 if (5>=x>=0)  
                                 else(80 if  (50>=x>=5) 
                                      else(60 if (500>=x>=50)
                                          else(40 if (10000>=x>=500) 
                                              else 0)))))
```

```python
#calulation of Ph
df['npH']=df.ph.apply(lambda x: (100 if (8.5>=x>=7)  
                                 else(80 if  (8.6>=x>=8.5) or (6.9>=x>=6.8) 
                                      else(60 if (8.8>=x>=8.6) or (6.8>=x>=6.7) 
                                          else(40 if (9>=x>=8.8) or (6.7>=x>=6.5)
                                              else 0)))))
```

```python
#calculation of dissolved oxygen
df['ndo']=df.do.apply(lambda x:(100 if (x>=6)  
                                 else(80 if  (6>=x>=5.1) 
                                      else(60 if (5>=x>=4.1)
                                          else(40 if (4>=x>=3) 
                                              else 0)))))
```

```python
#calc of B.O.D
df['nbdo']=df.bod.apply(lambda x:(100 if (3>=x>=0)  
                                 else(80 if  (6>=x>=3) 
                                      else(60 if (80>=x>=6)
                                          else(40 if (125>=x>=80) 
                                              else 0)))))
```

```python
#calculation of electrical conductivity
df['nec']=df.co.apply(lambda x:(100 if (75>=x>=0)  
                                 else(80 if  (150>=x>=75) 
                                      else(60 if (225>=x>=150)
                                          else(40 if (300>=x>=225) 
                                              else 0)))))
```

```python
#Calulation of nitrate
df['nna']=df.na.apply(lambda x:(100 if (20>=x>=0)  
                                 else(80 if  (50>=x>=20) 
                                      else(60 if (100>=x>=50)
                                          else(40 if (200>=x>=100) 
                                              else 0)))))
```

```python
df['wph']=df.npH * 0.165
df['wdo']=df.ndo * 0.281
df['wbdo']=df.nbdo * 0.234
df['wec']=df.nec* 0.009
df['wna']=df.nna * 0.028
df['wco']=df.nco * 0.281
```

```python
df['wqi']=df.wph+df.wdo+df.wbdo+df.wec+df.wna+df.wco 
```
